```python
# Pure relational SQL utilizing your clean, verified Silver schema headers
gold_query = """
SELECT
    sd.order_number,
    pr.product_key,
    cu.customer_key,
    sd.order_date,
    sd.shipping_date,
    sd.due_date,
    sd.sales_amount,
    sd.quantity,
    sd.price
FROM workspace.silver.crm_sales_details sd
LEFT JOIN workspace.gold.dim_products pr
    ON sd.product_number = pr.product_number
LEFT JOIN workspace.gold.dim_customers cu
    ON sd.customer_id = cu.customer_id
"""
df = spark.sql(gold_query)

# Save directly as a high-performance Gold Delta Table with atomic overwrite
df.write \
    .mode("overwrite") \
    .format("delta") \
    .option("overwriteSchema", "true") \
    .saveAsTable("workspace.gold.fact_sales")

# Display a clean data sample preview to verify integrity
df.display()
```
